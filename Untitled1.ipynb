```python
import time
import cv2
import queue
import threading
import tkinter as tk
from tkinter import ttk, messagebox
from datetime import datetime

from PIL import Image, ImageTk
from picamera2 import Picamera2
from ultralytics import YOLO
import RPi.GPIO as GPIO


# ================= CONFIGURATION =================

MODEL_PATH = "/home/wheelchair/Desktop/final/best_banana.pt"

SERVO_PIN = 4
SERVO_FREQUENCY = 50

CONFIDENCE_THRESHOLD = 0.60
STABLE_FRAMES = 3
SORT_DELAY = 1.0
WAIT_AFTER_SORT = 5.0

CAMERA_WIDTH = 640
CAMERA_HEIGHT = 480

# Chemical classification servo angles
FRUIT_ANGLES = {
    "Safe": 50,
    "Unsafe": 100,
    "Artificial": 140,
}

# =================================================


class FruitSorterGUI:

    def __init__(self, root):
        self.root = root
        self.root.title("Smart Chemical Classification & Sorting System")
        self.root.geometry("1200x760")
        self.root.minsize(950, 650)
        self.root.configure(bg="#101827")

        self.events = queue.Queue()
        self.frames = queue.Queue(maxsize=1)

        self.worker = None
        self.stop_event = threading.Event()
        self.running = False
        self.closing = False

        self.total_sorted = 0
        self.fruit_counts = {
            "Safe": 0,
            "Unsafe": 0,
            "Artificial": 0,
        }

        self.current_fruit = tk.StringVar(value="No item detected")
        self.confidence = tk.StringVar(value="0.0%")
        self.servo_angle = tk.StringVar(value="--")
        self.system_status = tk.StringVar(value="READY")
        self.camera_status = tk.StringVar(value="OFFLINE")
        self.process_status = tk.StringVar(
            value="Click Start Sorting to begin."
        )
        self.sorted_count = tk.StringVar(value="0")

        # Load YOLO classification model
        self.log("Loading YOLO classification model...")
        self.model = YOLO(MODEL_PATH)
        self.log("YOLO model loaded successfully.")

        # GPIO setup
        GPIO.setmode(GPIO.BCM)
        GPIO.setup(SERVO_PIN, GPIO.OUT)

        self.pwm = GPIO.PWM(SERVO_PIN, SERVO_FREQUENCY)
        self.pwm.start(0)

        self.build_ui()

        self.root.protocol("WM_DELETE_WINDOW", self.exit_app)
        self.root.after(50, self.process_events)
        self.root.after(30, self.update_camera_view)

    # ================= USER INTERFACE =================

    def build_ui(self):
        style = ttk.Style()
        style.theme_use("clam")

        style.configure(
            "Horizontal.TProgressbar",
            troughcolor="#253247",
            background="#22C55E",
        )

        # Header
        header = tk.Frame(self.root, bg="#172337", height=85)
        header.pack(fill="x")
        header.pack_propagate(False)

        tk.Label(
            header,
            text="   SMART CHEMICAL CLASSIFICATION SORTER",
            font=("Arial", 18, "bold"),
            bg="#172337",
            fg="white",
        ).pack(side="left", padx=15, pady=18)

        self.status_badge = tk.Label(
            header,
            text="● READY",
            font=("Arial", 11, "bold"),
            bg="#172337",
            fg="#4ADE80",
        )
        self.status_badge.pack(side="right", padx=25)

        tk.Label(
            self.root,
            text="YOLO CLASSIFICATION  |  RASPBERRY PI  |  IOT SEGREGATION",
            font=("Arial", 9),
            bg="#101827",
            fg="#94A3B8",
        ).pack(anchor="w", padx=20, pady=(10, 5))

        # Main layout
        body = tk.Frame(self.root, bg="#101827")
        body.pack(fill="both", expand=True, padx=16, pady=8)

        body.columnconfigure(0, weight=3)
        body.columnconfigure(1, weight=2)
        body.rowconfigure(0, weight=1)

        # LEFT: Camera panel
        left = tk.Frame(body, bg="#172337")
        left.grid(row=0, column=0, sticky="nsew", padx=(0, 8))

        tk.Label(
            left,
            text="LIVE CAMERA FEED",
            font=("Arial", 12, "bold"),
            bg="#172337",
            fg="white",
        ).pack(anchor="w", padx=15, pady=12)

        self.camera_label = tk.Label(
            left,
            text="Camera is offline\n\nStart sorting to activate the camera",
            font=("Arial", 13),
            bg="#0B1220",
            fg="#94A3B8",
            compound="center",
        )
        self.camera_label.pack(
            fill="both", expand=True, padx=12, pady=(0, 10)
        )

        self.camera_info = tk.Label(
            left,
            text="Waiting for camera...",
            font=("Arial", 9),
            bg="#172337",
            fg="#94A3B8",
        )
        self.camera_info.pack(anchor="w", padx=15, pady=(0, 12))

        # RIGHT: Detection details
        right = tk.Frame(body, bg="#101827")
        right.grid(row=0, column=1, sticky="nsew", padx=(8, 0))

        self.make_card(
            right, "CHEMICAL STATUS", self.current_fruit, "#4ADE80"
        )
        self.make_card(
            right, "MODEL CONFIDENCE", self.confidence, "#60A5FA"
        )
        self.make_card(
            right, "SERVO ANGLE", self.servo_angle, "#FBBF24"
        )
        self.make_card(
            right, "TOTAL ITEMS SORTED", self.sorted_count, "#C084FC"
        )

        # Process monitor
        process = tk.Frame(right, bg="#172337")
        process.pack(fill="x", pady=(0, 10))

        tk.Label(
            process,
            text="PROCESS MONITOR",
            font=("Arial", 11, "bold"),
            bg="#172337",
            fg="white",
        ).pack(anchor="w", padx=12, pady=10)

        self.process_label = tk.Label(
            process,
            textvariable=self.process_status,
            font=("Arial", 10),
            bg="#172337",
            fg="#4ADE80",
            wraplength=330,
            justify="left",
        )
        self.process_label.pack(anchor="w", padx=12, pady=(0, 10))

        self.progress = ttk.Progressbar(
            process,
            mode="indeterminate",
            style="Horizontal.TProgressbar",
        )
        self.progress.pack(fill="x", padx=12, pady=(0, 12))

        # Sorting statistics
        stats = tk.Frame(right, bg="#172337")
        stats.pack(fill="x", pady=(0, 10))

        tk.Label(
            stats,
            text="SORTING SUMMARY",
            font=("Arial", 11, "bold"),
            bg="#172337",
            fg="white",
        ).pack(anchor="w", padx=12, pady=10)

        self.count_labels = {}

        for fruit in FRUIT_ANGLES:
            row = tk.Frame(stats, bg="#172337")
            row.pack(fill="x", padx=12, pady=4)

            tk.Label(
                row,
                text=fruit,
                font=("Arial", 10),
                bg="#172337",
                fg="#CBD5E1",
            ).pack(side="left")

            count_label = tk.Label(
                row,
                text="0",
                font=("Arial", 10, "bold"),
                bg="#172337",
                fg="white",
            )
            count_label.pack(side="right")
            self.count_labels[fruit] = count_label

        # Controls
        controls = tk.Frame(self.root, bg="#101827")
        controls.pack(fill="x", padx=18, pady=10)

        self.start_button = tk.Button(
            controls,
            text="▶  START SORTING",
            command=self.start_sorting,
            bg="#16A34A",
            fg="white",
            activebackground="#15803D",
            activeforeground="white",
            font=("Arial", 11, "bold"),
            relief="flat",
            cursor="hand2",
            padx=20,
            pady=12,
        )
        self.start_button.pack(side="left", padx=(0, 10))

        self.stop_button = tk.Button(
            controls,
            text="■  STOP",
            command=self.stop_sorting,
            bg="#DC2626",
            fg="white",
            activebackground="#B91C1C",
            activeforeground="white",
            font=("Arial", 11, "bold"),
            relief="flat",
            cursor="hand2",
            padx=20,
            pady=12,
            state="disabled",
        )
        self.stop_button.pack(side="left", padx=5)

        tk.Button(
            controls,
            text="EXIT",
            command=self.exit_app,
            bg="#334155",
            fg="white",
            activebackground="#475569",
            activeforeground="white",
            font=("Arial", 10, "bold"),
            relief="flat",
            cursor="hand2",
            padx=20,
            pady=12,
        ).pack(side="right")

        # Activity log
        log_frame = tk.Frame(self.root, bg="#172337")
        log_frame.pack(fill="x", padx=18, pady=(0, 12))

        tk.Label(
            log_frame,
            text="ACTIVITY LOG",
            font=("Arial", 10, "bold"),
            bg="#172337",
            fg="white",
        ).pack(anchor="w", padx=10, pady=(8, 4))

        self.log_box = tk.Text(
            log_frame,
            height=5,
            bg="#0B1220",
            fg="#CBD5E1",
            insertbackground="white",
            font=("Consolas", 9),
            relief="flat",
            state="disabled",
            wrap="word",
        )
        self.log_box.pack(fill="x", padx=10, pady=(0, 10))

    def make_card(self, parent, title, variable, color):
        card = tk.Frame(parent, bg="#172337")
        card.pack(fill="x", pady=(0, 10))

        tk.Label(
            card,
            text=title,
            font=("Arial", 9, "bold"),
            bg="#172337",
            fg="#94A3B8",
        ).pack(anchor="w", padx=12, pady=(10, 2))

        tk.Label(
            card,
            textvariable=variable,
            font=("Arial", 17, "bold"),
            bg="#172337",
            fg=color,
        ).pack(anchor="w", padx=12, pady=(0, 10))

    # ================= LOGGING & UI EVENTS =================

    def log(self, message):
        timestamp = datetime.now().strftime("%H:%M:%S")

        if hasattr(self, "log_box"):
            self.log_box.configure(state="normal")
            self.log_box.insert(
                "end", f"[{timestamp}] {message}\n"
            )
            self.log_box.see("end")
            self.log_box.configure(state="disabled")
        else:
            print(f"[{timestamp}] {message}")

    def post(self, event, *values):
        self.events.put((event, values))

    def process_events(self):
        try:
            while True:
                event, values = self.events.get_nowait()

                if event == "status":
                    message, color = values
                    self.process_status.set(message)
                    self.process_label.configure(fg=color)

                elif event == "log":
                    self.log(values[0])

                elif event == "classification":
                    name, confidence, angle = values
                    self.current_fruit.set(name)
                    self.confidence.set(f"{confidence * 100:.1f}%")
                    self.servo_angle.set(f"{angle}°" if angle is not None else "--")

                elif event == "invalid_item":
                    name, confidence = values
                    self.current_fruit.set("Not a valid item")
                    self.confidence.set(f"{confidence * 100:.1f}%")
                    self.servo_angle.set("--")

                elif event == "no_item":
                    self.current_fruit.set("No item detected")
                    self.confidence.set("0.0%")
                    self.servo_angle.set("--")

                elif event == "camera":
                    self.camera_status.set(values[0])
                    self.camera_info.configure(
                        text=f"Camera status: {values[0]}"
                    )

                elif event == "count":
                    fruit, total = values
                    self.total_sorted = total
                    self.sorted_count.set(str(total))
                    self.count_labels[fruit].configure(
                        text=str(self.fruit_counts[fruit])
                    )

                elif event == "finished":
                    message = values[0]
                    self.running = False
                    self.start_button.configure(state="normal")
                    self.stop_button.configure(state="disabled")
                    self.status_badge.configure(
                        text="● READY", fg="#4ADE80"
                    )
                    self.progress.stop()
                    self.log(message)

                    if not self.closing:
                        self.process_status.set(message)
                        self.process_label.configure(fg="#4ADE80")
                        self.camera_info.configure(
                            text="Camera status: OFFLINE"
                        )
                        self.camera_label.configure(
                            image="",
                            text="Camera is offline\n\n"
                                 "Click Start Sorting to restart",
                        )
                        self.camera_label.image = None
                        self.current_fruit.set("No item detected")
                        self.confidence.set("0.0%")
                        self.servo_angle.set("--")

        except queue.Empty:
            pass

        if not self.closing:
            self.root.after(50, self.process_events)

    def update_camera_view(self):
        try:
            frame = self.frames.get_nowait()

            image = Image.fromarray(frame)
            image.thumbnail((700, 480))

            photo = ImageTk.PhotoImage(image=image)
            self.camera_label.configure(image=photo, text="")
            self.camera_label.image = photo

        except queue.Empty:
            pass

        if not self.closing:
            self.root.after(30, self.update_camera_view)

    def publish_frame(self, frame):
        try:
            self.frames.put_nowait(frame)
        except queue.Full:
            try:
                self.frames.get_nowait()
            except queue.Empty:
                pass

            try:
                self.frames.put_nowait(frame)
            except queue.Full:
                pass

    # ================= SERVO CONTROL =================

    def set_servo_angle(self, angle):
        angle = max(0, min(180, angle))
        duty = 2.5 + (angle / 180.0) * 10.0

        self.pwm.ChangeDutyCycle(duty)
        time.sleep(0.5)
        self.pwm.ChangeDutyCycle(0)

    def control_servo(self, category_name):
        angle = FRUIT_ANGLES.get(category_name)

        if angle is None:
            self.post("log", f"Unknown category: {category_name}")
            return False

        self.post(
            "status",
            f"Moving servo to {angle}° for {category_name}...",
            "#FBBF24",
        )
        self.post("log", f"{category_name} classified. Servo: {angle}°")

        self.set_servo_angle(angle)

        if self.stop_event.is_set():
            return False

        self.post("status", "Servo positioned. Releasing item...", "#FBBF24")
        time.sleep(SORT_DELAY)

        self.post("status", "Allowing item to fall into bin...", "#FBBF24")
        time.sleep(0.7)

        if self.stop_event.is_set():
            return False

        self.post("status", "Returning servo to home position (90°)...", "#60A5FA")
        self.set_servo_angle(90)

        self.fruit_counts[category_name] += 1
        self.total_sorted += 1

        self.post("count", category_name, self.total_sorted)
        self.post("log", f"{category_name} sorted successfully.")

        return True

    # ================= START / STOP =================

    def start_sorting(self):
        if self.running:
            return

        self.stop_event.clear()
        self.running = True

        self.start_button.configure(state="disabled")
        self.stop_button.configure(state="normal")
        self.status_badge.configure(
            text="● RUNNING", fg="#4ADE80"
        )

        self.progress.start(12)
        self.process_status.set("Initializing camera...")
        self.camera_info.configure(text="Camera status: STARTING")

        self.log("Starting chemical classification system...")

        self.worker = threading.Thread(
            target=self.sorting_worker,
            daemon=True,
        )
        self.worker.start()

    def stop_sorting(self):
        if not self.running:
            return

        self.stop_event.set()
        self.stop_button.configure(state="disabled")
        self.status_badge.configure(
            text="● STOPPING", fg="#FBBF24"
        )
        self.process_status.set(
            "Stopping safely. Current operation may finish first..."
        )
        self.log("Stop requested.")

    # ================= YOLO CLASSIFICATION + CAMERA WORKER =================

    def sorting_worker(self):
        camera = None
        camera_started = False
        message = "Sorting stopped."

        try:
            camera = Picamera2()

            config = camera.create_preview_configuration(
                main={
                    "size": (CAMERA_WIDTH, CAMERA_HEIGHT),
                    "format": "RGB888",
                }
            )

            camera.configure(config)
            camera.start()
            camera_started = True
            time.sleep(2)

            self.post("camera", "ONLINE")
            self.post("status", "Camera online. Scanning chemical safety...", "#4ADE80")
            self.post("log", "Camera started. YOLO classification active.")

            stable_name = None
            stable_count = 0
            waiting_for_clear = False

            while not self.stop_event.is_set():

                frame = camera.capture_array()

                # Run classification inference
                results = self.model(frame, verbose=False)

                detected_name = None
                best_confidence = 0.0

                if len(results) > 0 and results[0].probs is not None:
                    probs = results[0].probs
                    top1_idx = int(probs.top1)
                    best_confidence = float(probs.top1conf)
                    detected_name = str(self.model.names[top1_idx]).strip()

                # Check if confidence is valid and meets threshold
                is_above_threshold = best_confidence >= CONFIDENCE_THRESHOLD
                is_valid_category = detected_name in FRUIT_ANGLES

                # Draw classification results on video frame overlay
                display_frame = frame.copy()
                if detected_name and is_above_threshold:
                    if is_valid_category:
                        overlay_text = f"{detected_name}: {best_confidence:.2f}"
                        text_color = (0, 255, 0)
                    else:
                        overlay_text = f"Not a valid item: {best_confidence:.2f}"
                        text_color = (0, 0, 255)

                    cv2.putText(
                        display_frame,
                        overlay_text,
                        (20, 40),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.9, text_color, 2,
                    )

                self.publish_frame(display_frame)

                # Handle Blank Background / Low Confidence / Invalid Objects
                if detected_name is None or not is_above_threshold:
                    stable_name = None
                    stable_count = 0
                    waiting_for_clear = False

                    self.post("no_item")
                    self.post(
                        "status",
                        "Scanning items (No item detected)...",
                        "#60A5FA",
                    )
                elif not is_valid_category:
                    # Non-banana object detected
                    stable_name = None
                    stable_count = 0
                    waiting_for_clear = False

                    self.post("invalid_item", detected_name, best_confidence)
                    self.post(
                        "status",
                        f"Not a valid item ({detected_name}). Ignored.",
                        "#F87171",
                    )
                else:
                    angle = FRUIT_ANGLES.get(detected_name)

                    self.post(
                        "classification",
                        detected_name,
                        best_confidence,
                        angle,
                    )

                    if detected_name == stable_name:
                        stable_count += 1
                    else:
                        stable_name = detected_name
                        stable_count = 1

                    if not waiting_for_clear:
                        self.post(
                            "status",
                            f"Classified {detected_name}: "
                            f"{stable_count}/{STABLE_FRAMES} stable frames",
                            "#60A5FA",
                        )

                    if (
                        stable_count >= STABLE_FRAMES
                        and not waiting_for_clear
                        and angle is not None
                    ):
                        self.post(
                            "status",
                            f"Sorting {detected_name}...",
                            "#FBBF24",
                        )

                        # Pause camera during physical sorting.
                        camera.stop()
                        camera_started = False
                        self.post("camera", "PAUSED")
                        self.post("log", "Camera paused for sorting.")

                        sorted_ok = False

                        try:
                            sorted_ok = self.control_servo(detected_name)

                            if sorted_ok and not self.stop_event.is_set():
                                self.post(
                                    "status",
                                    "Sorting complete. Waiting before next item...",
                                    "#4ADE80",
                                )
                                time.sleep(WAIT_AFTER_SORT)

                        finally:
                            if not self.stop_event.is_set():
                                camera.start()
                                camera_started = True
                                time.sleep(1)
                                self.post("camera", "ONLINE")
                                self.post(
                                    "log",
                                    "Camera restarted.",
                                )

                        waiting_for_clear = True
                        stable_count = 0
                        stable_name = None

                        if not self.stop_event.is_set():
                            self.post(
                                "status",
                                "Remove the sorted item to detect the next one.",
                                "#60A5FA",
                            )

            message = "Sorting stopped safely."

        except Exception as exc:
            message = f"Error: {exc}"
            self.post("log", message)
            self.post("status", message, "#F87171")

        finally:
            if camera is not None:
                try:
                    if camera_started:
                        camera.stop()
                    camera.close()
                except Exception as exc:
                    self.post("log", f"Camera cleanup warning: {exc}")

            self.post("camera", "OFFLINE")
            self.post("finished", message)

    # ================= EXIT / CLEANUP =================

    def exit_app(self):
        if self.closing:
            return

        if self.running:
            if not messagebox.askyesno(
                "Exit Application",
                "Sorting is running. Stop the process and exit?",
                parent=self.root,
            ):
                return

            self.closing = True
            self.stop_event.set()
            self.start_button.configure(state="disabled")
            self.stop_button.configure(state="disabled")
            self.root.after(100, self.wait_for_worker)
        else:
            self.closing = True
            self.cleanup()
            self.root.destroy()

    def wait_for_worker(self):
        if self.worker and self.worker.is_alive():
            self.root.after(100, self.wait_for_worker)
        else:
            self.cleanup()
            self.root.destroy()

    def cleanup(self):
        try:
            self.set_servo_angle(90)
        except Exception as exc:
            print("Servo parking warning:", exc)

        try:
            self.pwm.stop()
            GPIO.cleanup()
        except Exception as exc:
            print("GPIO cleanup warning:", exc)


# ================= APPLICATION ENTRY =================

if __name__ == "__main__":
    root = tk.Tk()

    try:
        app = FruitSorterGUI(root)
        root.mainloop()

    except Exception as exc:
        print("Application startup error:", exc)
        try:
            GPIO.cleanup()
        except Exception:
            pass
        raise
```
